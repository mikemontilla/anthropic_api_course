## System Prompts

````python
# Adapting function to include system prompt
# This first approach doesn't work

def chat(messages, system_prompt=None):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages,
        system=system_prompt
    )
    return message.content[0].text

messages = []
system_prompt = """
You are a patient math tutor. 
Do not directly answer a student's question.
Patiently, guide them to a solution step by step
"""
add_user_message(messages, "What is the solution of 3x + 4 = 25") 
answer = chat(messages) # when no system prompt is provided the call fails
print(answer)
````

````python
# Adapting function to include system prompt
# This is the right solution

def chat(messages, system_prompt=None):

    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages
    }

    if system_prompt is not None:
        params["system"] = system_prompt
    
    message = client.messages.create(**params)
    return message.content[0].text

messages = []
system_prompt = """
You are a patient math tutor. 
Do not directly answer a student's question.
Patiently, guide them to a solution step by step
"""
add_user_message(messages, "What is the solution of 3x + 4 = 25") 
answer = chat(messages, system_prompt) # Works with or without system_prompt
print(answer)
````

# Temperature
Temperature controls how random or predictable Claude's responses are—lower values (near 0) make output more focused and deterministic, while higher values (up to 1) make it more varied and creative.

````python
def chat(messages, system_prompt=None, temperature=1.0):

    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature
    }

    if system_prompt is not None:
        params["system"] = system_prompt
    
    message = client.messages.create(**params)
    return message.content[0].text

messages = []
add_user_message(messages, "Generate a one sentence movie idea") 
answer = chat(messages, temperature=1.0) # More random responses (useful for creativity)
print(answer)
````

# Streaming
Streaming sends Claude's response back token by token as it's generated, rather than waiting for the full message to complete—so text appears incrementally, like watching it get typed in real time.

````python
messages = []
add_user_message(messages, "Write a 1 sentence description of a fake database")

# This time we receive a stream instead of a static response
stream = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages,
    stream=True
)

# stream is an iterable with multiple events
for event in stream:
    print(event)
````

Each event has a different meaning in the response, however there is a type of event which is more important because it has the parts of the real response: **RawContentBlockDeltaEvent**. We can check for this event or use the method available in the anthropic sdk...

````python
messages = []
add_user_message(messages, "Write a 1 sentence description of a fake database")

# This time the method is stream and not create, we don't need the stream parameter
with client.messages.stream(
    model=model,
    max_tokens=1000,
    messages=messages
) as stream:
    for text in stream.text_stream:
        print(text, end="") # to ensure no newline is added

# If we need the full message we can get it by doing this...
stream.get_final_message()
````

## Structured Data and Stop Sequences
Structured data lets you constrain Claude's output into a predictable format (like JSON) so it can be reliably parsed by your code, while stop_sequences are custom strings that tell Claude to halt generation as soon as it produces them.

````python
messages = []
add_user_message(messages, "Generate a very short example of json data structure")
answer = chat(messages)
print(answer)
````

We can see in the last example that if we need only the json this is not the ideal response. We can note however that backticks (```) are generated so it can be readable in a md file. We can use this to simulate part of the response of the assistant and a stop sequence to get only the json part

````python
def chat(messages, system_prompt=None, temperature=1.0, stop_sequences=[]):

    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature,
        "stop_sequences": stop_sequences
    }

    if system_prompt is not None:
        params["system"] = system_prompt
    
    message = client.messages.create(**params)
    return message.content[0].text


messages = []
add_user_message(messages, "Generate a very short example of json data structure")
#add_assistant_message(messages, "```json") # This is deprecated, the messages must end with a user message
answer = chat(messages, stop_sequences=["```"])
print(answer)
````

The current approach is the following... Using `output_config` property in which it is possible to specify the structure of the response:

````python
def chat(messages, system_prompt=None, temperature=1.0, output_config={}):

    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature,
        "output_config": output_config
    }

    if system_prompt is not None:
        params["system"] = system_prompt
    
    message = client.messages.create(**params)
    return message.content[0].text


messages = []
add_user_message(messages, "Generate a very short example of json data structure")
output_config = {
    "format": {
        "type": "json_schema",
        "schema": {
            "type": "object",
            "properties": {
                "name": {"type": "string"},
                "email": {"type": "string"},
            },
            "required": ["name", "email"],
            "additionalProperties": False,
        },
    }
}
answer = chat(messages, output_config=output_config)
print(answer)
````

````python
import json, re

response = client.messages.create(
    model="claude-opus-4-8",
    max_tokens=1024,
    temperature=1.0,  # higher temp = more varied/creative structures
    system="Respond with a single valid JSON value and nothing else. "
           "No markdown code fences, no commentary. You may choose whatever "
           "keys, nesting, and structure best fit the content.",
    messages=[{"role": "user", "content": "Invent a fantasy creature as JSON."}],
)

text = response.content[0].text.strip()
text = re.sub(r"^```(?:json)?|```$", "", text).strip()  # strip fences if any slip in
data = json.loads(text)  # wrap in try/except in production

print(text)
print("\n\n")
print(data)
````
